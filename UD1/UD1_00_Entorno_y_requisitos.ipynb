## 1. Instalar Python

Si ya tienes Python instalado (desde [https://www.python.org](https://www.python.org)),
puedes usar `pip` para instalar las librerías directamente:

## 2. Crear un entorno virtual (recomendado)

Trabajar con entornos virtuales evita conflictos entre versiones de librerías de
distintos proyectos.

### Con conda (si usas Anaconda)

```bash
conda create -n etcdfm python=3.11
conda activate etcdfm
```

### Con venv (si usas pip)

```bash
python -m venv etcdfm
# En Windows:
etcdfm\Scripts\activate
# En macOS / Linux:
source etcdfm/bin/activate
```

## 5. Verificar la instalación

Ejecuta la celda siguiente. Si todas las librerías están instaladas correctamente
verás un mensaje de confirmación para cada una. Si alguna falla, instálala con
`pip install <nombre_paquete>` y vuelve a ejecutar.


````python
paquetes = [
    ('pandas',       'pd'),
    ('numpy',        'np'),
    ('matplotlib',   'matplotlib'),
    ('sklearn',      'sklearn'),
    ('requests',     'requests'),
    ('dask',         'dask'),
    ('duckdb',       'duckdb'),
    ('networkx',     'nx'),
    ('pyarrow',      'pyarrow'),
]

print("Verificando paquetes instalados...")
print("-" * 45)
todos_ok = True
for nombre, alias in paquetes:
    try:
        mod = __import__(nombre)
        version = getattr(mod, '__version__', 'desconocida')
        print(f"  OK  {nombre:<15s} v{version}")
    except ImportError:
        print(f"  FALTA  {nombre:<15s}  ->  pip install {nombre}")
        todos_ok = False

print("-" * 45)
if todos_ok:
    print("Todo correcto. Puedes empezar con el Bloque 1.")
else:
    print("Instala los paquetes marcados como FALTA y vuelve a ejecutar.")
````
