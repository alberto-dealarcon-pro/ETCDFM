- crea la carpeta **`.venv`** dentro de `ETCDFM` (tu entorno virtual);
- instala las librerías del módulo **con las versiones exactas** fijadas en `uv.lock`,

Las librerías están declaradas en `pyproject.toml`. Los paquetes de la biblioteca estándar de Python
(`sqlite3`, `json`, `pathlib`, `datetime`, `xml`…) ya vienen con Python y no aparecen en la lista.

## 6. Abrir el proyecto y comprobar que todo funciona

1. Abre la carpeta en VS Code. Desde la terminal, dentro de `ETCDFM`:

   ```powershell
   code .
   ```

   (O en VS Code: **Archivo → Abrir carpeta…** y selecciona `ETCDFM`.)
   Si VS Code pregunta si confías en los autores de la carpeta, pulsa **Sí, confío en los autores**.

2. Abre este cuaderno: **`UD1/UD1_00_Entorno_y_requisitos.ipynb`**.

3. Arriba a la derecha pulsa **Select Kernel** (*Seleccionar kernel*) →
   **Python Environments** → **`.venv (Python 3.12.x)`**.

   > Si algún cuaderno te pide kernel, elige siempre **`.venv`**.

4. Ejecuta la celda siguiente (**Mayús + Intro**). Debes ver:
   - que el intérprete está en `ETCDFM\.venv\...`;
   - todas las librerías con **OK**;
   - el resultado de una pequeña prueba que usa pandas, Parquet, DuckDB y Dask a la vez.

````python
import sys
import importlib
from pathlib import Path

# 1) ¿Qué Python está ejecutando este cuaderno?
interprete = Path(sys.executable)
print(f"Python {sys.version.split()[0]}  ->  {interprete}")
if ".venv" in interprete.parts:
    print("  OK  El kernel es el entorno .venv del repositorio")
else:
    print("  AVISO  El kernel NO es .venv: cámbialo en 'Select Kernel' (arriba a la derecha)")

# 2) ¿Están todas las librerías del módulo?
paquetes = [
    # (módulo que se importa, nombre en pyproject.toml)
    ("pandas",     "pandas"),
    ("numpy",      "numpy"),
    ("pyarrow",    "pyarrow"),
    ("openpyxl",   "openpyxl"),
    ("lxml",       "lxml"),
    ("h5py",       "h5py"),
    ("PIL",        "pillow"),
    ("duckdb",     "duckdb"),
    ("sqlalchemy", "sqlalchemy"),
    ("dask",       "dask"),
    ("networkx",   "networkx"),
    ("requests",   "requests"),
    ("matplotlib", "matplotlib"),
    ("sklearn",    "scikit-learn"),
    ("ipywidgets", "ipywidgets"),
]

print("\nVerificando paquetes instalados...")
print("-" * 45)
faltan = []
for modulo, nombre in paquetes:
    try:
        mod = importlib.import_module(modulo)
        version = getattr(mod, "__version__", "")
        print(f"  OK     {nombre:<14s} {version}")
    except ImportError:
        print(f"  FALTA  {nombre}")
        faltan.append(nombre)
print("-" * 45)

# 3) Prueba rápida: pandas -> Parquet -> DuckDB y Dask
if not faltan:
    import tempfile
    import pandas as pd
    import duckdb
    import dask.dataframe as dd

    df = pd.DataFrame({"sensor": ["A", "B", "A", "C", "B"],
                       "valor":  [21.5, 19.0, 22.1, 18.4, 19.8]})
    with tempfile.TemporaryDirectory() as tmp:
        ruta = Path(tmp) / "prueba.parquet"
        df.to_parquet(ruta)
        resumen = duckdb.sql(
            f"SELECT sensor, round(avg(valor), 2) AS media FROM '{ruta.as_posix()}' "
            "GROUP BY sensor ORDER BY sensor"
        ).df()
    total_dask = dd.from_pandas(df, npartitions=2)["valor"].sum().compute()

    print(resumen.to_string(index=False))
    print(f"Suma calculada con Dask: {total_dask:.1f}")
    print("\nTodo correcto. Puedes empezar con el Bloque 1.")
else:
    print("Faltan librerías. Comprueba que el kernel es .venv y ejecuta `uv sync` en la terminal.")
````

> **Si ves la tabla con las medias por sensor, tu entorno está listo.**

## 10. Solución de problemas

| Problema | Solución |
|---|---|
| `git`, `uv` o `code` *"no se reconoce como nombre de un cmdlet…"* | Cierra **todas** las terminales (y VS Code) y abre una nueva. Si sigue, revisa el `PATH` (paso 3) o reinicia el ordenador. |
| El script de instalación de `uv` da error de *directiva de ejecución* | Usa la alternativa `winget install --id astral-sh.uv -e`. |
| `uv sync` no puede descargar Python (red del centro, proxy…) | Instala **Python 3.12** desde <https://www.python.org/downloads/> y vuelve a ejecutar `uv sync`: lo detectará. |
| En *Select Kernel* no aparece `.venv` | Comprueba que abriste la carpeta **`ETCDFM`** (no una superior). Pulsa `Ctrl + Mayús + P` → **Python: Select Interpreter** → `.venv`. Si no está, ejecuta `uv sync` y recarga la ventana (`Ctrl + Mayús + P` → **Developer: Reload Window**). |
| `ModuleNotFoundError: No module named '...'` | Estás en otro kernel. Cambia a `.venv`. Si ya lo estás, ejecuta `uv sync`. |
| VS Code pide instalar `ipykernel` | No lo instales desde el aviso: significa que el kernel no es `.venv`. Cámbialo. |
| El entorno se ha estropeado | Cierra VS Code, borra la carpeta `.venv` dentro de `ETCDFM` y ejecuta `uv sync`. Se vuelve a crear en unos minutos. |
| `git pull` da conflicto con un cuaderno | Ver el apartado 7, *Trabajo del día a día*. |
