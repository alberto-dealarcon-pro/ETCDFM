## Ejercicio 2.3 — Pokémon en CSV, JSON y Parquet

En el ejercicio 1.2 construiste un DataFrame de Pokémon. Ahora vamos a ampliar ese dataset a **100 Pokémon** y a comparar los tres formatos de almacenamiento más habituales.

**Parte A:** descarga los datos y guárdalos en los tres formatos.


````python
import requests
import pandas as pd
import time
import os

# Descarga de 100 Pokemon (puede tardar unos segundos)
registros = []
for i in range(1, 101):
    datos = requests.get(f'https://pokeapi.co/api/v2/pokemon/{i}').json()
    tipos = [t['type']['name'] for t in datos['types']]
    stats = {s['stat']['name']: s['base_stat'] for s in datos['stats']}
    registros.append({
        'id':       datos['id'],
        'nombre':   datos['name'],
        'peso':     datos['weight'],
        'altura':   datos['height'],
        'tipo_1':   tipos[0],
        'tipo_2':   tipos[1] if len(tipos) > 1 else None,
        'hp':       stats.get('hp', 0),
        'ataque':   stats.get('attack', 0),
        'defensa':  stats.get('defense', 0),
        'velocidad': stats.get('speed', 0),
    })

df = pd.DataFrame(registros)
print(f'Dataset: {len(df)} Pokemon, {len(df.columns)} columnas')
print(df.head(3))
````

````python
import time
import os

# Guardar en los tres formatos
df.to_csv('pokemon.csv', index=False)
df.to_json('pokemon.json', orient='records', indent=2)
df.to_parquet('pokemon.parquet', index=False)

# Comparar tamaños
formatos = ['pokemon.csv', 'pokemon.json', 'pokemon.parquet']
print('Comparativa de tamaño en disco:')
for f in formatos:
    size = os.path.getsize(f)
    print(f'  {f:<20s}: {size:>8,} bytes ({size/1024:.1f} KB)')
````

**Parte B:** compara los tiempos de lectura y responde las preguntas.


````python
import time

def medir_lectura(func, nombre):
    t0 = time.perf_counter()
    df = func()
    t1 = time.perf_counter()
    print(f'  {nombre:<25s}: {(t1-t0)*1000:.2f} ms')
    return df

print('Tiempo de lectura completa:')
medir_lectura(lambda: pd.read_csv('pokemon.csv'),           'CSV')
medir_lectura(lambda: pd.read_json('pokemon.json'),         'JSON')
medir_lectura(lambda: pd.read_parquet('pokemon.parquet'),   'Parquet')

# Lectura con filtro (solo tipo fuego)
print('\nLectura filtrando tipo_1 == fuego:')
medir_lectura(lambda: pd.read_csv('pokemon.csv').query("tipo_1 == 'fire'"),       'CSV + filtro pandas')
medir_lectura(lambda: pd.read_parquet('pokemon.parquet', filters=[('tipo_1','=','fire')]), 'Parquet + filtro pushdown')
````

## Ejercicio 3.2 — Dask con la Pokédex completa

En el ejercicio 2.3 trabajaste con 100 Pokémon. Ahora vamos a descargar los **primeros 500** y procesarlos con Dask para notar la diferencia respecto a pandas.

**Parte A:** descarga los datos (puede tardar un par de minutos) y guárdalos en Parquet para no tener que repetir la descarga.


````python
import requests
import pandas as pd
from pathlib import Path

parquet_path = 'pokedex_500.parquet'

if not Path(parquet_path).exists():
    print('Descargando 500 Pokemon...')
    registros = []
    for i in range(1, 501):
        datos = requests.get(f'https://pokeapi.co/api/v2/pokemon/{i}').json()
        tipos = [t['type']['name'] for t in datos['types']]
        stats = {s['stat']['name']: s['base_stat'] for s in datos['stats']}
        registros.append({
            'id':         datos['id'],
            'nombre':     datos['name'],
            'peso':       datos['weight'],
            'altura':     datos['height'],
            'tipo_1':     tipos[0],
            'tipo_2':     tipos[1] if len(tipos) > 1 else None,
            'hp':         stats.get('hp', 0),
            'ataque':     stats.get('attack', 0),
            'defensa':    stats.get('defense', 0),
            'sp_ataque':  stats.get('special-attack', 0),
            'sp_defensa': stats.get('special-defense', 0),
            'velocidad':  stats.get('speed', 0),
        })
    pd.DataFrame(registros).to_parquet(parquet_path, index=False)
    print('Guardado en', parquet_path)
else:
    print('Fichero ya existe, cargando...')

df_pandas = pd.read_parquet(parquet_path)
print(f'Dataset: {len(df_pandas)} filas, {len(df_pandas.columns)} columnas')
````

**Parte B:** realiza las mismas operaciones con pandas y con Dask, y compara.


````python
import dask.dataframe as dd
import time

df_dask = dd.from_pandas(df_pandas, npartitions=4)

print('Particiones de Dask:', df_dask.npartitions)
print('Tipo de df_dask:', type(df_dask))
print()

# Operacion 1: media de estadisticas por tipo
t0 = time.perf_counter()
res_pandas = df_pandas.groupby('tipo_1')[['hp','ataque','defensa','velocidad']].mean().round(1)
t1 = time.perf_counter()
print(f'Pandas  — media por tipo: {(t1-t0)*1000:.2f} ms')
print(res_pandas.sort_values('ataque', ascending=False).head(5))

t0 = time.perf_counter()
res_dask = df_dask.groupby('tipo_1')[['hp','ataque','defensa','velocidad']].mean().compute().round(1)
t1 = time.perf_counter()
print(f'Dask    — media por tipo: {(t1-t0)*1000:.2f} ms')
````

Usando el dataset de 500 Pokémon, calcula cuántos Pokémon hay de cada tipo con la lógica MapReduce: primero emite pares `(tipo, 1)` para cada Pokémon, luego agrupa por tipo y suma.


````python
from collections import defaultdict

# ── FASE MAP ─────────────────────────────────────────────────────────────
# Emitir (tipo_1, 1) para cada Pokemon
def map_tipo(pokemon):
    return (pokemon['tipo_1'], 1)

pares_mapeados = [map_tipo(row) for row in df_pandas.to_dict('records')]
print('Primeros 5 pares mapeados:', pares_mapeados[:5])

# ── FASE SHUFFLE (agrupar por clave) ──────────────────────────────────────
grupos = defaultdict(list)
for clave, valor in pares_mapeados:
    grupos[clave].append(valor)

# ── FASE REDUCE ──────────────────────────────────────────────────────────
def reduce_suma(clave, valores):
    return (clave, sum(valores))

resultado = dict(reduce_suma(k, v) for k, v in grupos.items())
resultado_ordenado = sorted(resultado.items(), key=lambda x: x[1], reverse=True)

print('\nConteo de Pokemon por tipo (MapReduce):')
for tipo, cantidad in resultado_ordenado:
    print(f'  {tipo:<12}: {cantidad}')
````
