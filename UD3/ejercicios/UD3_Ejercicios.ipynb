# El caso: los gimnasios municipales de Granada

> Lee esta descripción **con atención y entera** antes de empezar. Todo lo que necesitas para diseñar la base de datos y para resolver los ejercicios está aquí, y algunos detalles importan más de lo que parece.

El Área de Deportes del Ayuntamiento de Granada gestiona la red de gimnasios municipales **«Granada en Forma»**, con **seis centros**: Zaidín, Chana, Albaicín, Cartuja, Genil y Almanjáyar. Cada centro tiene un nombre, una dirección, un barrio y un horario de apertura (todos abren de 7:00 a 22:00 de lunes a sábado; el domingo solo abren de 9:00 a 14:00). Cada centro está dividido en **varias salas** (musculación, cardio, actividades dirigidas y, en dos de ellos, piscina), y **el aforo máximo se fija para cada sala**, no para el centro. El centro de Albaicín **cierra todo el mes de agosto** por mantenimiento.

Cualquier persona de 16 años o más puede hacerse **abonada**. Al darse de alta se registran su fecha de nacimiento, su sexo, el barrio en el que vive, la fecha de alta y el **centro habitual** que elige, aunque puede entrar en cualquier centro de la red. Por protección de datos, el sistema de análisis no guarda el nombre ni el DNI: cada abonado se identifica con un número.

Hay **cuatro tipos de abono**, cada uno con una cuota mensual:

| Abono | Condiciones | Cuota mensual |
|-------|-------------|:-------------:|
| General | Sin restricciones | 34,90 € |
| Joven | Hasta los 25 años | 24,90 € |
| Mayor | Desde los 65 años | 19,90 € |
| Mañanas | Solo puede entrar antes de las 15:00 | 22,90 € |

Cada vez que un abonado contrata un abono se crea un **contrato**, con su fecha de inicio. Un abonado puede cambiar de tipo de abono (se cierra el contrato anterior y se abre otro) y puede darse de baja y, tiempo después, volver: el abonado es el mismo, pero el contrato es nuevo.

**Bajas.** La baja voluntaria se **solicita** en recepción o en la web y se hace **efectiva el último día del mes siguiente** al de la solicitud (un mes de preaviso): durante ese tiempo el abonado sigue pagando y puede seguir entrando. Además, cuando un abonado acumula **tres recibos devueltos**, el sistema le da de **baja de oficio** por impago, sin que haya solicitud. En cada baja se guarda su motivo.

**Accesos.** Cada centro tiene un **torno** que el abonado pasa con su tarjeta. El torno registra **tanto la entrada como la salida**, con la fecha y la hora y el sentido del paso. Los tornos guardan los pasos en el ordenador del propio centro y, cada noche, cada centro **exporta un fichero CSV** con los pasos del día que se carga en el servidor central del Ayuntamiento.

**Cobros.** El primer día de cada mes se emite un **recibo** por cada contrato activo, que se cobra por domiciliación bancaria. Unos días después el banco envía un fichero con los recibos que no ha podido cobrar, que pasan a estado **devuelto**. Los recibos del mes en curso que aún no se han procesado están **pendientes**.

**Lo que quiere el Ayuntamiento.** El servicio pierde muchos abonados y quiere actuar **antes** de que se vayan: cada día 1 de mes, quiere una lista de los abonados con más riesgo de **solicitar la baja voluntaria** en los 30 días siguientes, para llamarles y ofrecerles algo. Las bajas por impago no le interesan para esta campaña.

El periodo de datos que vas a simular va del **1 de enero de 2024 al 31 de diciembre de 2025**.

````python
import os
import numpy as np
import pandas as pd
from sqlalchemy import create_engine, text, event

DB = 'gimnasio.db'

def crear_engine_sqlite(ruta):
    # Engine de SQLite que activa las claves ajenas en CADA conexión que abre
    eng = create_engine(f'sqlite:///{ruta}')

    @event.listens_for(eng, 'connect')
    def _activar_fk(dbapi_con, _registro):
        dbapi_con.execute('PRAGMA foreign_keys = ON')

    return eng

# Partimos siempre de una base de datos vacía
if os.path.exists(DB):
    os.remove(DB)
engine = crear_engine_sqlite(DB)

DDL_MI_DISENO = '''
-- TODO: escribe aquí el DDL de tu diseño (CREATE TABLE ..., CREATE INDEX ...)
'''

with engine.begin() as conn:
    for sentencia in DDL_MI_DISENO.split(';'):
        if sentencia.strip() and not sentencia.strip().startswith('--'):
            conn.execute(text(sentencia))

# TODO: lista las tablas e índices creados y sus claves ajenas
````

# Bloque 4 — El dataset para predecir el abandono

## La fecha de corte

Para entrenar un modelo que prediga el abandono necesitamos una tabla con **una fila por ejemplo**, unas **variables** (lo que sabemos del abonado) y una **etiqueta** (lo que queremos predecir). La pieza clave para construirla es la **fecha de corte** (*cutoff date* o *snapshot date*).

La fecha de corte es el instante que hace de **«presente»** en cada fila del dataset:

- todo lo que pasó **antes** de la fecha de corte sirve para calcular las **variables**;
- lo que pasa **después** solo sirve para calcular la **etiqueta**.

**Ejemplo.** Tomamos como fecha de corte el **1 de marzo de 2025** para un abonado:

```
            pasado (variables)                  │      futuro (etiqueta)
 ─────────────────────────────────────────────── ┼ ──────────────────────────
  enero: 12 visitas · febrero: 4 visitas          │  ¿solicita la baja entre el
  1 recibo devuelto · 2 años de antigüedad...     │  1 y el 30 de marzo?
                                         1 de marzo de 2025
```

- **Variables:** visitas en las 4 semanas anteriores al 1 de marzo, días desde su última visita hasta esa fecha, recibos devueltos hasta esa fecha, antigüedad a esa fecha…
- **Etiqueta:** 1 si ocurre lo que queremos predecir en los 30 días siguientes al 1 de marzo, y 0 si no.

Así reproducimos la situación real en la que se usará el modelo: el día 1 de cada mes el gimnasio solo sabe lo que ha ocurrido hasta ese día y quiere anticipar lo que pasará en el mes.

### La fuga de datos

Si alguna variable usa información **posterior** a la fecha de corte, se produce **fuga de datos** (*data leakage*). Por ejemplo, unas «visitas totales del abonado» calculadas con toda la tabla `accesos` incluirían las visitas de marzo. El modelo parecerá excelente al evaluarlo, pero fallará en producción, porque ese dato todavía no existe el 1 de marzo. El caso extremo es usar como variable una columna que contiene la propia respuesta, como la fecha de baja.

### Varias fechas de corte

Normalmente no se usa una sola fecha de corte, sino **una por mes** (1 de febrero, 1 de marzo, 1 de abril…). Cada abonado activo en cada fecha de corte genera una fila, de modo que:

- hay muchos más ejemplos para entrenar;
- un mismo abonado aparece varias veces, con su situación en cada momento;
- se puede entrenar con los cortes antiguos y evaluar con los más recientes, como pasaría en la realidad.

Es la misma idea que el *point-in-time join* del Bloque 3 de teoría (`merge_asof` con `direction='backward'`): a cada momento le corresponden las variables calculadas **antes**, nunca después.
