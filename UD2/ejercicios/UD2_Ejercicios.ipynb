## Ejercicio 2.3 — De CSV a Excel con múltiples hojas

Retomamos el dataset de sismos del ejercicio anterior. Un geólogo te pide un fichero Excel con:

- **Hoja 1** (`Todos`): todos los sismos con las columnas `time`, `latitude`, `longitude`, `depth`, `mag`, `place`.
- **Hoja 2** (`Fuertes`): solo los de magnitud ≥ 6.0, ordenados de mayor a menor magnitud.
- **Hoja 3** (`Resumen`): tabla con el número de sismos y magnitud media agrupados por `magType`.

El Excel debe tener la primera fila congelada y la columna de magnitud resaltada en amarillo en la hoja `Fuertes`.


````python
import pandas as pd
from openpyxl.styles import PatternFill

# Asegúrate de tener df_sismos cargado del ejercicio anterior
# Si no, descárgalo de nuevo aquí

COLS = ['time', 'latitude', 'longitude', 'depth', 'mag', 'place']

# TODO: crea df_todos, df_fuertes y df_resumen
# df_todos   = ...
# df_fuertes = ...
# df_resumen = ...

# Exportar a Excel con múltiples hojas
excel_path = 'sismos_informe.xlsx'
df_todos = df_sismos[COLS].copy()

# Excel no admite fechas con zona horaria
df_todos['time'] = pd.to_datetime(df_todos['time'], utc=True).dt.tz_localize(None)

df_fuertes = (
    df_todos[df_todos['mag'] >= 6.0]
    .sort_values('mag', ascending=False)
)

df_resumen = (
    df_sismos.groupby('magType', dropna=False)
    .agg(
        numero_sismos=('mag', 'size'),
        magnitud_media=('mag', 'mean')
    )
    .reset_index()
)

with pd.ExcelWriter(excel_path, engine='openpyxl') as writer:
    df_todos.to_excel(writer, sheet_name='Todos', index=False)
    df_fuertes.to_excel(writer, sheet_name='Fuertes', index=False)
    df_resumen.to_excel(writer, sheet_name='Resumen', index=False)

    amarillo = PatternFill(fill_type='solid', fgColor='FFFF00')

    for ws in writer.book.worksheets:
        ws.freeze_panes = 'A2'

    ws = writer.book['Fuertes']
    col_mag = [c.column for c in ws[1] if c.value == 'mag'][0]

    for row in ws.iter_rows(
        min_row=2,
        min_col=col_mag,
        max_col=col_mag
    ):
        row[0].fill = amarillo

print(f'Fichero guardado: {excel_path}')

# Verificación: lee el Excel de vuelta y muestra las hojas
import openpyxl
wb = openpyxl.load_workbook(excel_path)
print(f'Hojas: {wb.sheetnames}')
````
